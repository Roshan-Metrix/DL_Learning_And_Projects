## Simple Neural Network

```python
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np
```

```python
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data() 
```

```python
# Scaling to get better Accuracy
X_train = X_train / 255
X_test = X_test / 255
```

```python
X_train[0]
```

```python
# Converting 2-D Array to 1-D Array
X_train_flattened = X_train.reshape(len(X_train),28*28)
X_test_flattened = X_test.reshape(len(X_test),28*28)
```

```python
X_test_flattened.shape
```

```python
# Without hidden layer , only input and output
model = keras.Sequential([
    keras.layers.Dense(10,input_shape=(784,),activation='sigmoid') # 784 input & 10 output
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.fit(X_train_flattened, y_train, epochs=5)
```

```python
model.evaluate(X_test_flattened, y_test)
```

### Let's try image 

```python
# First Image
plt.matshow(X_test[0])
```

```python
y_predicted = model.predict(X_test_flattened)
y_predicted[0] # first image
```

```python
np.argmax(y_predicted[0]) # Taking Greater value from Output
```

```python
# Hence it's 7 
```

```python
# Finding all predicted value
y_predicted_labels = [np.argmax(i) for i in y_predicted]
y_predicted_labels[:5]
```

```python
y_test[:5]
```

# Confusion Matrix

```python
cm = tf.math.confusion_matrix(labels=y_test,predictions=y_predicted_labels)
cm
```

```python
import seaborn as sn
plt.figure(figsize = (10,7))
sn.heatmap(cm, annot=True, fmt='d')
plt.xlabel('Predicted')
plt.ylabel('Truth')
```

```python
# Adding hidden layer
model = keras.Sequential([
    keras.layers.Dense(100,input_shape=(784,),activation='relu'), # this layer has 784 input & 100 neurons
    keras.layers.Dense(10, activation='sigmoid') # this layer have 10 neurons
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.fit(X_train_flattened, y_train, epochs=5)
```

```python
model.evaluate(X_test_flattened, y_test)
```

```python
y_predicted = model.predict(X_test_flattened)
y_predicted_labels = [np.argmax(i) for i in y_predicted]
```

```python
cm = tf.math.confusion_matrix(labels=y_test,predictions=y_predicted_labels)
cm
```

```python
import seaborn as sn
plt.figure(figsize = (10,7))
sn.heatmap(cm, annot=True, fmt='d')
plt.xlabel('Predicted')
plt.ylabel('Truth')
```
